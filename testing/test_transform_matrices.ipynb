```python
# --- Example: Matrix size N scales automatically with wavelet and max_level ---
N = 16

wavelet_names = ['haar', 'db2', 'db3', 'db4', 'coif1', 'sym2', 'sym4']

test_wavelets = {
    name: Wavelet_matrix(vector_length=N, wavelet=name)
    for name in wavelet_names
}
```

```python
for key, value in test_wavelets.items():
    print(key)
    plt.imshow(value.transform_matrix.T,cmap='gray')
    plt.title(key)
    plt.show()
```
